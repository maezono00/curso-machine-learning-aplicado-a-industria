Usando o Python nativo para gerar uma lista de números.

```python
#Definindo o tamanho que a lista vai ter.
n = 1000000
```

```python
lista = list(range(n))
```

```python
#Registrando ponto de partida para o cálculo do tempo usado para gerar a lista.
t0 = time.time()
```

Gerando uma lista de 1 milhão de números.

```python
resultado = [x**2 for x in lista]
```

```python
#Registrando o ponto final e fazendo o cálculo do tempo que levou para gerar a lista.
t1 = time.time()
t1 - t0
```

```python
# Cria o array e calcula o quadrado diretamente de forma vetorizada
t0 = time.time()
arr = np.arange(n)
resultado = arr ** 2
t1 = time.time()

tempo_execucao = t1 - t0
print(f"Tempo: {tempo_execucao:.5f} segundos")
```
