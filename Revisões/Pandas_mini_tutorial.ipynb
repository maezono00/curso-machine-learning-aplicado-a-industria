```python
#iloc serve para Localizar por linha
df.iloc[0, :]
```

```python
#Exibindo as colunas da posição 1 e 2
df.iloc[:, 1:3]
```

```python
#Executando groupby
g = df.groupby('risco')
g['renda'].mean()
g['renda'].agg(['mean', 'std', 'count'])
g.size()
```

```python
#Verificando quantos valores ausentes existem
df.isnull().sum()
```

```python
#Excluir linhas com valores nulos
df.dropna()
```

```python
#Preencher valores nulos da coluna renda usando a média.
df['renda'].fillna(df['renda'].median())
```

```python
#O método ffill vai usar como base o valor anterior e jogará na célula seguinte para não deixar vazia.
df.fillna(method = 'ffill')
```

Otimizando memória convertendo alguns dados para tipos menores

Inicialmente, todos os campos do DataFrame estavam como `float64`, acontece que alguns dados não precisam estar em `float64`, até porque, por exemplo, não existe uma cama e meia.

```python
california_housing.info()
```

```python
california_housing['population'].describe()
```

Alterando o tipo do campo `population` para `int16`, levando em consideração que o `int16` aceita até 65555 inserções de dados (além de que não existe uma pessoa e meia, então outro motivo pra não ser necessário o uso do `float64`).

```python
california_housing['population'] = california_housing['population'].astype('int16')
```

```python
california_housing.info()
```

Alterando o campo `households` para `int16` pelo mesmo motivo, não existe 1 casa e meia.

Fazendo o teste com o `.info()` junto do `.astype('int16')` para verificar quanto de memória iria ser consumida.

```python
california_housing['households'].astype('int16').info()
```

```python
california_housing['households'] = california_housing['households'].astype('int16')
```

Após essas duas alterações, houve uma melhora considerável de memória, porém, ainda há dados que dariam pra ser convertidos para `int`.

```python
california_housing.info()
```

```python
#Consultando o uso da memória após algumas otimizações.
california_housing.memory_usage()
```

Dando continuidade nas otimizações, o campo `total_bedrooms` também foi convertido para `int16`.

```python
california_housing['total_bedrooms'] = california_housing['total_bedrooms'].astype('int16')
```

```python
california_housing.info()
```

As duas últimas colunas convertidas foram `total_rooms` e `housing_median_age`, dessas colunas que foram alteradas a única que foi alterada pra `int8` foi `housing_median_age` levando em consideração que o comprimento do campo não passava de 255 caracteres.

```python
california_housing['total_rooms'] = california_housing['total_rooms'].astype('int16')
```

```python
california_housing['housing_median_age'] = california_housing['housing_median_age'].astype('int8')
```

```python
california_housing.info()
```

```python
california_housing.memory_usage()
```

Relatório da diferença entre o dataframe bruto e o dataframe otimizado.
> Foram feitas otimizações nos campos `households`, `population`, `housing_medium_age`, `total_bedrooms` e `total_rooms`. Os campos citados originalmente estavam como `float`, agora estão como `int`, assim, exigindo menos memória.

```python
california_bruto = pd.read_csv('/content/sample_data/california_housing_train.csv')
california_antes = california_bruto.memory_usage().sum()
```

```python
california_depois = california_housing.memory_usage().sum()
```

```python
print(f"Antes: {california_antes} bytes\nDepois: {california_depois} bytes\nA economia foi de: {(((california_antes - california_depois) / california_antes) * 100):.2f}%")
```
