## Atividade de fixação 1.7

Crie uma lista com pelo menos quatro produtos. Cada produto deve ser um dicionário contendo:

- nome;
- categoria;
- preço;
- estoque;
- situação ativa ou inativa.

Depois:

1. mostre o nome do segundo produto;
2. altere o estoque do primeiro;
3. adicione um novo produto;
4. crie um conjunto com as categorias sem repetição;
5. calcule o total de unidades em estoque.

````python
#Criando uma lista de pelo menos quatro produtos.

lista_produtos = [
    {"nome": "Banana", "categoria": "Fruta", "preco": 3.99, "estoque": 100, "situacao": "Ativo"},
    {"nome": "Batata", "categoria": "Legume", "preco": 1.99, "estoque": 5, "situacao": "Ativo"},
    {"nome": "Alface", "categoria": "Verdura", "preco": 5, "estoque": 50, "situacao": "Ativo"},
    {"nome": "Cenoura", "categoria": "Legume", "preco": 4.99, "estoque": 25, "situacao": "Ativo"},
    {"nome": "Tomate", "categoria": "Fruta", "preco": 2.99, "estoque": 0, "situacao": "Inativo"}
]
````

Foi gerada uma lista com 5 produtos, para exibir o nome do segundo produto é feito o seguinte:
- é chamada a `lista_produtos[1]`, sendo `lista_produtos` o nome da lista e o `[1]` a segunda posição da lista `(As listas começam com índice 0)`.
- junto do `lista_produtos[1]` é acrescentado o `["nome"]`, esse `["nome"]` vai estar chamando qual categoria será exibida.
- ou seja, em `lista_produtos[1]["nome"]` vai chamar o que está registrado na "coluna" `nome` na segunda posição da lista.

````python
#Exibindo o nome do segundo produto.
print(lista_produtos[1]["nome"])
````

Para alterar algum item existente de uma lista é feito um procedimento bem parecido quando vamos exibir uma coluna de algum item da lista.
> Foi feito o exercício de alterar o estoque do primeiro item da lista
- Primeiro, chamamos a lista, a posição e a coluna que queremos editar
  ```python
  lista_produtos[0]["estoque"]
  ```
- Após isso, usando o sinal de `=`, atribuímos o novo valor desse item:
  ```python
  lista_produtos[0]["estoque"] = 67
  ```
- Sendo assim, o produto que está presente no índice 0 terá um estoque de 67 unidades.

````python
#Alterar estoque do primeiro produto.
lista_produtos[0]["estoque"] = 67

print(f"O estoque do produto: {lista_produtos[0]["nome"]} é de {lista_produtos[0]["estoque"]} unidades.")
````

##1.2 Atividade de fixação
1. Crie uma lista de dicionários com cinco produtos.
2. Salve a lista em `produtos.json`.
3. Leia o arquivo novamente.
4. Mostre apenas produtos ativos.
5. Gere também um arquivo `produtos.csv`.
6. Trate o caso em que o arquivo não existe.

````python
#Criando a lista de dicionários com 5 produtos.
produtos = [
    {"nome": "Mouse", "preco": 80.0, "estoque": 10, "ativo": True},
    {"nome": "Teclado", "preco": 100.0, "estoque": 20, "ativo": False},
    {"nome": "Mousepad", "preco": 150.0, "estoque": 100, "ativo": False},
    {"nome": "Carregador", "preco": 25.5, "estoque": 5, "ativo": True},
    {"nome": "Monitor", "preco": 500.99, "estoque": 25, "ativo": True}
]
````

###Fazendo a operação de gravar a lista dos produtos em um arquivo json.
> A formatação do comando open é assim: nome do arquivo, operação desejada (se será feita a leitura ou gravação do arquivo), definir o encoding do arquivo.

````python
#Importando a biblioteca json.
import json

#O "w" no comando open vem de "write".
with open("produtos.json", "w", encoding = "utf-8") as arquivo:
  json.dump(
      produtos,
      arquivo,
      ensure_ascii = False,
      indent = 2
  )
````

###Importando o arquivo gerado no ambiente.
> A estrutura para ser feita a leitura de um arquivo é bem parecida, é usada a função open e dentro dela é inserida o nome do arquivo, o comando `"r"` para ser feita a leitura e o enconding.

````python
with open("produtos.json", "r", encoding = "utf-8") as arquivo:
  produto_carregado = json.load(arquivo)

#Leitura do arquivo carregado.
print(produto_carregado)
````

````python
#Mostrar apenas produtos ativos.
print("Lista de produtos ativos:")
for produto in produto_carregado:
  if produto['ativo'] == True:
    print(f"{produto['nome']}")
````

###Gerando um arquivo .csv
> Foi feita a adaptação da base de dados para o formato csv, ou seja, a primeira linha deve ser as colunas que serão usadas na base de dados e logo em seguida são as informações, respeitando as ordens das colunas citadas na primeira linha.

Novamente, foi usado o comando `open()` e atribuindo todo esse conteúdo para a variável `arquivo`, assim, fazendo todas as operações necessárias para o mesmo.

````python
import csv

produtos = [
    ["nome", "preco", "estoque", "ativo"],
    ["Mouse", 80.0, 10, True],
    ["Teclado", 100.0, 20, False],
    ["Mousepad", 150.0, 100, False],
    ["Carregador", 25.5, 5, True],
    ["Monitor", 500.99, 25, True]
]

with open("produtos.csv", "w", newline = "", encoding = "utf-8") as arquivo:
  escritor = csv.writer(arquivo)
  escritor.writerows(produtos)
````

###Fazendo a leitura e exibição do arquivo .csv gerado

````python
with open("produtos.csv", "r", encoding = "utf-8") as arquivo:
  leitor = csv.DictReader(arquivo)

  for linha in leitor:
    print(linha)
````

###Tratativa de erros caso o arquivo ainda não exista.
> Foi pedido que fosse feita a leitura de um arquivo não existente no projeto, chamado `configuracao.json`. Como o arquivo não existe, foram feitas as tratativas de erros.

````python
from json.decoder import JSONDecodeError
try:
  with open("configuracao.json", "r", encoding = "utf-8") as arquivo:
    configuracao = json.load(arquivo)
#Tratativa para caso o arquivo não seja encontrado.
except FileNotFoundError:
  configuracao = {}
  print("O arquivo ainda não existe, uma configuração vazia foi criada!")
#Tratativa para caso o JSON possua um formato inválido.
except JSONDecodeError:
  configuracao = {}
  print("O arquivo JSON possui um formato inválido.")
````

##1.3 Análise de vendas com Pandas
Crie um DataFrame ou leia um arquivo contendo as seguintes informações:

- nome do produto;
- categoria do produto;
- preço unitário;
- quantidade vendida;
- custo unitário.

Em seguida, realize as etapas abaixo:

1. visualize as primeiras linhas do DataFrame com `head()`;
2. verifique a estrutura, os tipos de dados e a presença de valores nulos com `info()`;
3. obtenha um resumo estatístico das colunas numéricas com `describe()`;
4. crie a coluna `faturamento`, calculada por:
    
    `preço × quantidade vendida`;
    
5. crie a coluna `custo_total`, calculada por:
    
    `custo unitário × quantidade vendida`;
    
6. crie a coluna `lucro_bruto`, calculada por:
    
    `faturamento − custo_total`;
    
7. filtre apenas os produtos que apresentaram lucro bruto positivo;
8. ordene os produtos do maior para o menor faturamento;
9. agrupe os dados por categoria e apresente o faturamento, o custo total e o lucro bruto de cada grupo;
10. salve o resultado final em um arquivo CSV.

O arquivo CSV deve ser salvo sem a coluna de índice.

````python
#Criação e exibição do dataframe.
import pandas as pd

produtos = [
    {"nome_produto": "Mouse", "categoria_produto": "Periférico", "preco_unitario": 10, "qtd_vendida": 100, "custo_unitario": 5},
    {"nome_produto": "Torraedira", "categoria_produto": "Eletrodoméstico", "preco_unitario": 100, "qtd_vendida": 50, "custo_unitario": 50},
    {"nome_produto": "Celular", "categoria_produto": "Eletrônico", "preco_unitario": 3000, "qtd_vendida": 200, "custo_unitario": 2000}
]

df = pd.DataFrame(produtos)
df
````

````python
#Exibição das primeiras linhas do DataFrame.
df.head()
````

````python
#Exibição da estrutura, tipos de dados e presença de valores nulos no DataFrame.
df.info()
````

````python
#Resumo estatístico do DataFrame.
df.describe()
````

````python
#Criação da coluna "Faturamento" usando como base o Preço Unitário dos produtos e a Quantidade Vendida.
df["faturamento"] = df["preco_unitario"] * df["qtd_vendida"]

#Exibição do DataFrame após essa operação de criação da coluna.
df
````

````python
#Criação da coluna "custo_total" com base nas colunas "custo_unitario" e "quantidade_vendida"
df["custo_total"] = df["custo_unitario"] * df["qtd_vendida"]

#Exibição do resultado no DataFrame.
df
````

````python
#Criação da coluna "lucro_bruto" com base na diferença do "faturamento" com o "custo_total".
df["lucro_bruto"] = df["faturamento"] - df["custo_total"]

#Exibição do DataFrame após as alterações.
df
````

````python
#filtre apenas os produtos que apresentaram lucro bruto positivo;
bruto_positivo = df [df["lucro_bruto"] > 1000]

print(bruto_positivo)
````
