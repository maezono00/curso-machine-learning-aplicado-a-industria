Fazendo as tratativas para preencher valores NaN.
> É aconselhado que para valores números seja preenchido com a mediana e valores categóricos sejam preenchidos com moda.

Preenchendo os valores NaN da coluna `operador` usando a moda (Valor com maior ocorrência) presente no dataset
> Será atribuído na coluna `operador` a coluna `operador` com as alterações desejadas
  - Nesse exemplo foi aplicado o `.fillna()`, dentro desse `.fillna()` foi feito a busca de qual era a moda da coluna `operador`, com o comando `.mode()` e, vendo a lista de modas da coluna, foi optado por usar a moda de índice 0.

```python
df['operador'] = df['operador'].fillna(df['operador'].mode()[0])
```

```python
df['modo_controle'] = df['modo_controle'].fillna(df['modo_controle'].mode()[0])
```

É possível fazer uma lista com as colunas que possuem valores nulos e fazer um laço de repetição para que seja feita todo esse processo em cada coluna dessa lista.
> No exemplo abaixo foi uma lista com as colunas categóricas, dessa lista foi feito um laço de repetição para que, cada coluna dessa lista que precise executar o `.fillna()`, seja feito a etapa de encontrar a moda e preencher com o primeiro item (índice 0) da lista de modas (que é o item que mais se repete).

```python
categorias = ['lote_producao', 'operador', 'modo_controle', 'status_valvula']

for categoria in categorias:
  df[categoria] = df[categoria].fillna(df[categoria].mode()[0])
```

Todas as colunas categóricas atualmente possuem 0 linhas vazias.

```python
df.isna().sum()
```

Para facilitar o processo de separar as colunas numéricas para as categóricas, pode-se usar o comando `.select_dtypes()`, dentro do parênteses será usado um `include = ` com a categoria desejada, por fim, inserir `.columns`, gerando uma lista que possua apenas o nome das colunas.
> Exemplo `df.select_dtypes(include = np.number).columns`<br>
> Caso queira que seja feito a lista das categórias, pode-se usar o `exclude` no lugar do `include`, assim, excluindo as colunas numéricas da lista de colunas exibidas.

```python
df.select_dtypes(include = np.number).columns
```

Repetindo o mesmo processo para as colunas numéricas, preenchendo os campos `NaN` com a mediana.

```python
categorias_numericas = ['setpoint_vazao_l_min', 'vazao_medida_l_min', 'pressao_linha_bar', 'temperatura_processo_c', 'abertura_valvula_pct', 'vibracao_bomba_mm_s', 'tempo_resposta_ms', 'ph_produto', 'densidade_produto_g_cm3']

for categoria in categorias_numericas:
  df[categoria] = df[categoria].fillna(df[categoria].median())
```

---
Calculando o IQR Manualmente

```python
q1 = df['pressao_linha_bar'].quantile(0.25)
```

```python
q3 = df['pressao_linha_bar'].quantile(0.75)
```

```python
iqr = q3 - q1
```

```python
lim_inf = q1 - 1.5 * iqr
```

```python
lim_sup = q3 + 1.5 * iqr
```

```python
print(f"Q1 (25%): {q1:.2f}\nQ3 (75%): {q3:.2f}\nIQR: {iqr:.2f}\nLimite Inferior: {lim_inf:.2f}\nLimite Superior: {lim_sup:.2f}")
```

```python
df['pressao_linha_bar'].clip(lower = lim_inf, upper = lim_sup)
```
