Operadores booleanos (condicionais) comparações.

````python
idade = int(input("Digite a sua idade: "))

#Verifica a primeira condição.
if idade >= 18:
  print("Você é maior de idade, pode tirar carteira de motoristas!")
#Se a primeira condição for falsa, verificará a segunda condição.
elif idade >= 16:
  print("Você tem idade para votar, mas não para dirigir.")
#Se nenhuma das anteriores for verdadeira.
else:
  print("Você é menor de idade.")
````

Em `or`, dos condicionais booleanos, é preciso que pelo menos uma das condições seja verdadeira para que o resultado seja `True`!

````python
idade = 30
sexo = "M"
print((idade == 30) or (sexo == "F"))
````

Em `and`, dos condicionais booleanos, é preciso que as duas condições sejam verdadeiras para que o resultado seja `True`!

````python
idade = 30
sexo = "F"

print((idade == 30) and (sexo == "M"))
````

Caso sejam usados mais de um condicional booleano em uma mesma sentença, é preciso se atentar na ordem de operação desses condicionais.
> Nesse exemplo, primeiro veio o `and` e depois o `or`. O valor do `and` foi `False`, já que somente uma das condições era a verdadeira. Agora para o `or`, deu `True`, pois ele precisa que pelo menos das sentenças seja `True` (que nesse caso, é `aluno = "True"`).

````python
idade = 30
sexo = "M"
aluno = "True"

print((idade == 20) and (sexo == "M") or (aluno))
````

>A verificação não necessita parênteses, no entanto, por questão de boas práticas é interessante que seja usado.

````python
idade = 10
sexo = "M"
aluno = "True"

print(idade == 10 and sexo == "F" and aluno == 10)
````

Quando um conjunto de dados está dentro de Colchetes (`[]`), é um(a) array / matriz / lista, ou seja, é um conjunto de dados que pode ser alterado!<br>
Quando esse conjunto de dados está dentro de parênteses (`()`), é uma tupla, ou seja, o conjunto desses dados não pode ser alterado. Muito usado quando precisamos manter a integridade dos dados.

````python
alimentos = ['arroz', 'queijo', 'feijão', 'salada', 'carne', 'tomate']
nome = ('Heitor', 'Oliveira')
````

Para consultar um item que esá dentro de uma lista, independente se for matriz ou uma tupla, é sempre usado a ordem:
1. Nome da lista
2. Posição do item dentro da lista em colchetes (`[]`)
3. Caso essa lista de itens seja tipo um json, inserir o nome da "coluna" desejada entre colchetes (`[]`)

O resultado será algo tipo:
```python
lista[0]
```

````python
alimentos[4]
````

````python
nome[0]
````

> É possível fazer alterações em listas mutáveis de várias formas, uma delas é atribuindo um valor em um índice específico.

````python
alimentos[0] = 'café'
````

> Já em listas imutáveis (como as tuplas) **não** é possível fazer alterações! (mesmo fazendo todo o trabalho de indicar qual o índice e atribuir um valor nele)

````python
nome[0] = 'Batista'
````

> A forma simples de imprimir/exibir os elementos presentes dentro de uma lista, é só chamando essa lista e executar. A resposta serão os itens disponíveis na lista, porém, formatado em forma de array/matriz

````python
alimentos
````

Uma das formas de imprimir os itens dentro de uma lista é usando o laço de repetição `for`
> Nesse exemplo criamos a variável `alimento`, ela irá receber cada item presente na lista `alimentos` e, ao fazer o `print()`, exibirá o item correspondente.

````python
for alimento in alimentos:
  print(alimento)
````

É possível também criar condicionais para a exibição desses itens.
> Caso o item seja `café` ou `queijo`, será exibido no texto como `Para o café temos:`. Caso contrário, `Para o almoço temos:`.

````python
for alimento in alimentos:
  if ((alimento == "café") or (alimento == "queijo")):
    print(f"Para o café temos: {alimento}")
  else:
    print(f"Para o almoço temos: {alimento}")
````

Outra opção, é criar uma lista dentro da condicional e pedir que seja feita a verificação dos itens especificados.
> Nesse exemplo, caso os itens analisados estejam dentro da lista da condicional, será exibida na seguinte mensagem:

````python
for alimento in alimentos:
  if alimento in ['café', 'queijo']:
    print(f"Para o café, temos: {alimento}.")
````

> Caso não esteja na lista especificada, o item será exibido na seguinte mensagem:

````python
for alimento in alimentos:
  if alimento not in ['café', 'queijo']:
    print(f"Para o almoço, temos: {alimento}.")
````

#Laço de repetição WHILE
Um outro laço de repetição disponível é o `while`, ele já é um pouco mais específico, algo do tipo "enquanto essa condição não for `True`, a tarefa será executada".
> Por exemplo, um programa que só vai parar de ser executado ao digitar `sair`. Sendo satisfeita essa condição, o programa vai encerrar a execução.
>> Quando é feita uma comparação entre strings, qualquer diferença importa, para evitar dores de cabeça do programa não encerrar ao digitar `SAIR` ou `Sair`, durante a comparação é usado a função/modificador de string `.lower()`, assim indepente da forma que o usuário digitar, será reconhecido no sistema como um texto em letras minúsculas.

````python
comando = 'continue'

while comando.lower() != 'sair':
  comando = input('Digite um comando: ')
  print(f"Você digitou o comando: {comando}")
````

Outras opções de tratamento das Strings:

````python
ferramenta  = 'fErrO dE sOldA'

print(f"Converter a String para caixa alta: {ferramenta.upper()}.")
print(f"Converter a String para letras minúsculas: {ferramenta.lower()}.")
print(f"Converter as primeiras letras de cada palavra em maiúsculo (muito usado em nomes próprios): {ferramenta.title()}.")
print(f"Converter apenas a primeira letra da String para maiúsculo: {ferramenta.capitalize()}.")
````

#Slicing de Strings
Outra possibilidade de manipulação dos textos é o `Slicing`, muito usado em tabelas.
Para executarmos o `Slicing` é feito o seguinte:
- Uma variável já deve ter a String desejada
- Ao ser feito o print, indicamos de qual index até qual index queremos exibir.

> Exemplo:
```python
texto_desejado = "Olá, Mundo!"
print(texto_desejado[index_inicial, index_final])
```

````python
texto = "Olá, Mundo!"
print(texto[2:7])
````

Outro exemplo de Slicing, é exibir os números de intervalos específicos.
> Por exemplo, exibir as letras num intervalo de 2 em 2 números.
```python
print(texto_desejado[::2])
```

````python
print(texto[::2])
````

Gerar uma lista de números um intervalo específico.
> É usado o comando `range()`, dentro os parênteses `()` é inserido o intervalo desejado.

````python
numeros = [n*2 for n in range(10)]
numeros
````

> Outra opção de intervalo de números: Uma lista que os números são uma divisão exata por `3` (Usando o operador módulo `%`)

````python
numeros = [n for n in range(10) if not n%3]
numeros
````

#Funções
Para a criação de funções, usamos o seguinte formato:
- `def`, para avisar que estamos criando uma função
- O nome da função
- As variáveis da função usadas, entre parênteses `()`
- Um return ou as instruções do que essa função faz.
> Exemplo:
  ```python
  def nome_da_funcao(variavel_1, variavel_2):
    return variavel1 - variavel2
  ```

````python
def subtracao(x, y):
  return x - y

def soma(x, y):
  return x + y

def divisao(x, y):
  resultado = x / y
  return resultado

def multiplicacao(x, y):
  resultado = x * y
  return resultado
````

````python
x = float(input("Insira o primeiro valor: "))
y = float(input("Insira o segundo valor: "))
````

````python
z = round(subtracao(x,y), 6)

print(f"O resulta da subtração entre {x} e {y} é: {z}")
````

````python
z = round(soma(x, y), 6)

print(f"O resultado da soma entre {x} e {y} é: {z}")
````

````python
print(f"O resultado da divisão entre {x} e {y} é: {round(divisao(x, y), 6)}")
````

````python
print(f"O resultado da multiplicação entre {x} e {y} é: {round(multiplicacao(x, y), 6)}")
````

#Manipulação de arquivos
Ao importar/fazer leitura de arquivos, é sempre importante atribuir essa ação de leitura em uma variável.

````python
f = open("demofile.txt", 'a')
````

````python
f.close()
````

> Outra opção para fazer a leitura dos arquivos, pois ao fazer a conversão para String, toda a estrutur de Lista não existe mais, portanto, é possível fazer a leitura em formato de lista com a biblioteca `ast`.

````python
import ast
````

> Leitura do arquivo, dando o alias de `arquivo` para fazer as operações e atribuindo todo o conteúdo em uma variável chamada `conteudo_string`.

````python
with open("alunos.txt", "r", encoding = "utf-8") as arquivo:
  conteudo_string = arquivo.read()
````

> Toda a operação de converter o conteúdo do arquivo para uma lista é feita pela função `.literal_eval()` da biblioteca `ast` e o resultado dessa operação é atribuída em uma variável.
```python
insira_variavel_para_lista = ast.literal_eval(insira_documento)
```

````python
dados = ast.literal_eval(conteudo_string)
````

> Verificação se a conversão funcionou.

````python
print(f"Tipo de variável: {type(dados)}")
print(f"Total de registros: {len(dados)}")
````

> Exibição dos dados desejados da lista.

````python
print(f"Nome: {dados[0]['nome']}")
print(f"Telefones: {dados[0]['telefone']}")
````
