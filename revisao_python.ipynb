###1.1.2 Outros exemplos de célula de código.

####A. Exemplo de operações atribuindo valores nas variáveis.

```python
#Atribuição dos valores numéricos nas variáveis a, b e c.
a = 10
b = 20
c = 30
```

```python
#Executando a soma de a, b e c e atribuindo o resultado dentro da variável soma.
soma = a + b + c
```

```python
#Exibindo o resultado da variável soma junto de um texto usando o formatador "f".
print(f"Ai que legal!!!! {soma}")
```

```python
#Executando a divisão entre os valores das variáveis a e b.
a / b
```

```python
#Exibindo, dentro de um print, o resultado da divisão entre a variável soma (60) divido pelo valor dentro da variável c (30).
print(soma / c)
```

```python
#Outro exemplo de operações com variáveis, dessa vez, somando A e B e, desse valor, subtraindo com o valor presente em C.
a + b - c
```

```python
#Criação da função somar, exemplificando também a importância da identação no código.
def somar(x, y):
  return x + y
```

```python
#Exemplo da execução da função somar, onde chamamos a função somar e dentro dos parênteses inserimos os dois valores que queremos somar separados por vírgula.
somar(10,20)
```

```python
#Exemplo de função somar, só que ao invés de fazer diretamente a soma, o valor dessa soma é atribuído à uma variável (resultado_soma) e logo após é feita a exibição do resultado.
def somar2(x, y):
  resultado_soma = x + y
  return resultado_soma
```

```python
#Execução da função somar2.
somar2(40,60)
```

```python
#Exemplo de que é possível também, caso a variável já exista, usar dentro da operação de uma função (Por exemplo: Pedi que a função "somar" fizesse a soma de 40 + o valor presente dentro da variável A).
somar(40, a)
```

```python
#Exemplo de multiplicação entre variáveis.
a * b
```

```python
#Exemplo de divisão entre variáveis.
b / a
```

```python
#Exemplo usando comparação de igualdade, nesse exemplo foi comparado se o valor dentro da variável A é igual à 10.
if a == 10:
  print("O valor da variável A é igual à 10.")
else:
  print("o valor da variável A não é igual à 10.")
```

```python
idade = int(input("Digite a sua idade: "))

#Verifica a primeira condição.
if idade >= 18:
  print("Você é maior de idade, pode tirar carteira de motoristas!")
#Se a primeira condição for falsa, verificará a segunda condição.
elif idade >= 16:
  print("Você tem idade para votar, mas não para dirigir.")
#Se nenhuma das anteriores for verdadeira.
else:
  print("Você é menor de idade.")
```
